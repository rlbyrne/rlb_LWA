```python
import numpy as np
import pyuvdata
import matplotlib.pyplot as plt
```

```python
use_file_old = "/lustre/pipeline/slow/46MHz/2025-10-06/19/20251006_192330_46MHz.ms"
use_file_new = "/lustre/pipeline/slow/46MHz/2025-10-06/19/20251006_195706_46MHz.ms"
use_file_after = "/lustre/pipeline/slow/46MHz/2025-10-06/20/20251006_202411_46MHz.ms"

uv_old = pyuvdata.UVData()
uv_old.read_ms(use_file_old)
uv_new = pyuvdata.UVData()
uv_new.read_ms(use_file_new)
uv_after = pyuvdata.UVData()
uv_after.read_ms(use_file_after)
```

```python
uv_new.select(antenna_names=["LWA257"])
uv_new.select(ant_str="auto")

uv_old.select(antenna_names=["LWA257"])
uv_old.select(ant_str="auto")

uv_after.select(antenna_names=["LWA257"])
uv_after.select(ant_str="auto")
```

```python
plt.plot(uv_new.freq_array/1e6, uv_new.data_array[0,:,0], ".-", markersize=1, linewidth=0.2, label="19:57:06 UTC")
plt.plot(uv_old.freq_array/1e6, uv_old.data_array[0,:,0], ".-", markersize=1, linewidth=0.2, label="19:23:30 UTC")
plt.plot(uv_after.freq_array/1e6, uv_after.data_array[0,:,0], ".-", markersize=1, linewidth=0.2, label="20:24:11 UTC")
plt.title("LWA-257 XX")
plt.ylabel("Autocorrelation Power")
plt.xlabel("Frequency (MHz)")
plt.legend()
plt.show()
```
