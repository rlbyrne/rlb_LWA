```python
import os
import numpy as np
import pathlib
```

```python
use_freq_bands = [
    "13",
    "18",
    "23",
    "27",
    "32",
    "36",
    "41",
    "46",
    "50",
    "55",
    "59",
    "64",
    "69",
    "73",
    "78",
    "82",
]

concatenated_freq_bands = [
    "34",
    "44",
    "52",
    "62",
    "72",
    "79",
    "83",
]
```

```python
all_dates = []
for freq in use_freq_bands:
    all_dates.extend(os.listdir(f"/lustre/pipeline/cosmology/{freq}MHz"))
for freq in concatenated_freq_bands:
    all_dates.extend(os.listdir(f"/lustre/pipeline/cosmology/concatenated_data/{freq}MHz"))
all_dates = np.sort(list(set(all_dates)))
```

```python
len([date for date in all_dates if date.startswith("2026")])
```

```python
orig_dir = "/lustre/pipeline/cosmology"
for date in all_dates:
    data_paths = []
    for freq in use_freq_bands:
        # Get files in a nested file structure
        data_paths.extend(
            [
                str(p)
                for p in pathlib.Path(f"{orig_dir}/{freq}MHz/{date}").rglob("*.ms")
                if p.is_dir()
            ]
        )
        data_paths.extend(
            [
                str(p)
                for p in pathlib.Path(f"{orig_dir}/{freq}MHz/{date}").rglob("*.ms.tar")
                if p.is_dir()
            ]
        )
    for freq in concatenated_freq_bands:
        data_paths.extend(
            [
                str(p)
                for p in pathlib.Path(f"{orig_dir}/concatenated_data/{freq}MHz/{date}").rglob("*.ms")
                if p.is_dir()
            ]
        )
        data_paths.extend(
            [
                str(p)
                for p in pathlib.Path(f"{orig_dir}/concatenated_data/{freq}MHz/{date}").rglob("*.ms.tar")
                if p.is_dir()
            ]
        )
    print(f"{date}: {len(data_paths)} data files.")
```
